```python
import pandas as pd
from scipy import stats


data = pd.read_excel(r'F:\Glomap_dNBR.xlsx',sheet_name='Sheet1')


data = data.dropna()


pre_dNBR = data['pre-2010-dNBR']
post_dNBR = data['post-2010-dNBR']

# Mann-Whitney U test
stat, p_value = stats.mannwhitneyu(pre_dNBR, post_dNBR, alternative='less')  # alternative='two-sided' 


print(f"Mann-Whitney U test: {stat}")
print(f"p value: {p_value}")


alpha = 0.05
if p_value < alpha:
    print("Mann-Whitney，significant")
else:
    print("Mann-Whitney，not significant")
```
